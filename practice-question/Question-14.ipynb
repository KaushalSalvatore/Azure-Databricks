- Given Input

+---+
|id |
+---+
| 1 |
| 2 |
| 3 |
+---+

- output 

+---+
|id |
+---+
| 1 |
| 2 |
| 2 |
| 3 |
| 3 |
| 3 |
+---+

```python
from pyspark.sql import SparkSession
from pyspark.sql import functions as F
```

```python
spark = SparkSession.builder \
    .appName("RepeatID") \
    .getOrCreate()
```

```python
df = spark.createDataFrame([(1,), (2,), (3,)], ["id"])
```

```python
result = df.withColumn(
    "repeat_col",
    F.sequence(F.lit(1), F.col("id"))
).select(
    "id",
    F.explode("repeat_col").alias("dummy")
).select("id")
```

```python
result.show()
```
