#### We have a dataset of different individuals and the items they possess along with their weights. Our goal is to aggregate this data so that we can see the total weight of each item for every person.

#### Input Data
data = [
    ("alice", "carrot", 1),
    ("bob", "banana", 3),
    ("alice", "tomato", 4),
    ("bob", "carrot", 2),
    ("charlie", "banana", 5),
    ("bob", "apple", 1),
    ("alice", "tomato", 2),
    ("charlie", "carrot", 3)
]

#### Solution 
- We first get the total weight for each item for each person
- We just collect the items for each person

```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import sum, collect_list, struct
```

```python
spark = SparkSession.builder.appName("PersonItemWeight").getOrCreate()
```

```python
data = [
    ("alice", "carrot", 1),
    ("bob", "banana", 3),
    ("alice", "tomato", 4),
    ("bob", "carrot", 2),
    ("charlie", "banana", 5),
    ("bob", "apple", 1),
    ("alice", "tomato", 2),
    ("charlie", "carrot", 3)
]
```

```python
cols = ["name", "item", "weight"]
```

```python
df = spark.createDataFrame(data, cols)
```

```python
aggregated_df = df.groupBy("name", "item").agg(sum("weight").alias("total_weight"))
```

```python
final_result = aggregated_df.groupBy("name").agg(
    collect_list(struct("item", "total_weight")).alias("item_list")
)
```

```python
final_result.show(truncate=False)
```
