```python
data = [(1,'abhay','male'),(2,'vivan','male'),(3,'Arav','male'),(4,'aman','male'),(5,'elena','female')]
cols = ['id','name','gender']
df3 = spark.createDataFrame(data,cols)
display(df3)
```

```python
df3.createOrReplaceTempView('persons')
```

```python
%sql
CREATE WIDGET DROPDOWN name DEFAULT 'vivan' CHOICES SELECT name from persons;
```

```python
%sql
CREATE WIDGET DROPDOWN name DEFAULT 'male' CHOICES SELECT DISTINCT gender from persons;
```

```python
%sql
select*from persons where gender = "female"
```
